```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyArrowPatch
from pathlib import Path

PROJECT_DIR = Path(r"C:\Users\annam\Dissertation 2026")
FIGURE_DIR = PROJECT_DIR / "figures" / "phase2_clustering_v2"

# ----------------------------
# Analytical pipeline diagram — which data state feeds which
# downstream method, generic across both datasets
# ----------------------------
fig, ax = plt.subplots(figsize=(17, 13))
ax.set_xlim(0, 17)
ax.set_ylim(0, 13)
ax.axis("off")

def box(x, y, w, h, text, facecolor="#E8F4F3", edgecolor="#0E7C7B", fontsize=10, fontweight="normal"):
    rect = mpatches.FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.05",
                                     facecolor=facecolor, edgecolor=edgecolor, linewidth=1.5)
    ax.add_patch(rect)
    ax.text(x + w/2, y + h/2, text, ha="center", va="center", fontsize=fontsize,
             fontweight=fontweight)

def arrow(x1, y1, x2, y2, color="black", lw=1.3):
    arr = FancyArrowPatch((x1, y1), (x2, y2), arrowstyle="-|>", mutation_scale=13,
                           color=color, linewidth=lw)
    ax.add_patch(arr)

# ---- Top: source object ----
box(6.5, 11.5, 4.0, 1.0, "Annotated object\n(post-QC, clustered, cell_type)",
    facecolor="#0F2A3D", edgecolor="#0F2A3D", fontsize=11, fontweight="bold")
ax.texts[-1].set_color("white")

# ---- Four data states branching off ----
states_y = 9.8
box(0.3, states_y, 3.0, 1.2, "Raw integer counts\n(clean_rawcounts.h5ad,\npre-normalisation)", facecolor="#FBF2E3", edgecolor="#B8862E", fontsize=9)
box(3.8, states_y, 3.0, 1.2, "Log-normalised,\nfull gene set\n(.raw)", facecolor="#FBF2E3", edgecolor="#B8862E", fontsize=9)
box(7.3, states_y, 3.0, 1.2, "Cell-type labels +\nsample IDs only\n(.obs, no expression)", facecolor="#FBF2E3", edgecolor="#B8862E", fontsize=9)
box(10.8, states_y, 3.0, 1.2, "Harmony-corrected\nembedding, per\npopulation subset", facecolor="#FBF2E3", edgecolor="#B8862E", fontsize=9)
box(14.1, states_y, 2.6, 1.2, "HVG-scaled,\nPCA embedding\n(clustering only)", facecolor="#FBF2E3", edgecolor="#B8862E", fontsize=9)

for x in [1.8, 5.3, 8.8, 12.3, 15.4]:
    arrow(x, 11.5, x, states_y + 1.2)

# ---- Analyses, one per state ----
an_y = 7.6
box(0.3, an_y, 3.0, 1.0, "Pseudobulk DE\n(PyDESeq2)", facecolor="#E8F4F3", edgecolor="#0E7C7B", fontsize=10)
box(3.8, an_y, 3.0, 1.0, "Marker validation\n+ LIANA", facecolor="#E8F4F3", edgecolor="#0E7C7B", fontsize=10)
box(7.3, an_y, 3.0, 1.0, "scCODA\n(compositional)", facecolor="#E8F4F3", edgecolor="#0E7C7B", fontsize=10)
box(10.8, an_y, 3.0, 1.0, "Sub-clustering", facecolor="#E8F4F3", edgecolor="#0E7C7B", fontsize=10)
box(14.1, an_y, 2.6, 1.0, "Leiden clustering", facecolor="#E8F4F3", edgecolor="#0E7C7B", fontsize=10)

for x in [1.8, 5.3, 8.8, 12.3, 15.4]:
    arrow(x, states_y, x, an_y + 1.0)

# ---- Outputs ----
out_y = 5.8
box(0.3, out_y, 3.0, 1.0, "DE gene lists\n(per comparison)", facecolor="#FFFFFF", edgecolor="#5C6B6E", fontsize=9)
box(3.8, out_y, 3.0, 1.0, "Confirmed cell\nidentities +\ninteraction networks", facecolor="#FFFFFF", edgecolor="#5C6B6E", fontsize=9)
box(7.3, out_y, 3.0, 1.0, "Credible\ncompositional\neffects", facecolor="#FFFFFF", edgecolor="#5C6B6E", fontsize=9)
box(10.8, out_y, 3.0, 1.0, "cell_type_fine\nsub-states", facecolor="#FFFFFF", edgecolor="#5C6B6E", fontsize=9)
box(14.1, out_y, 2.6, 1.0, "cell_type\n(parent labels)", facecolor="#FFFFFF", edgecolor="#5C6B6E", fontsize=9)

for x in [1.8, 5.3, 8.8, 12.3, 15.4]:
    arrow(x, an_y, x, out_y + 1.0)

# ---- One more step: DE -> pathway enrichment ----
arrow(1.8, out_y, 1.8, 4.4)
box(0.3, 3.4, 3.0, 1.0, "Pathway enrichment\n(gseapy, Hallmark)", facecolor="#E8F4F3", edgecolor="#0E7C7B", fontsize=10)

plt.tight_layout()
plt.savefig(FIGURE_DIR / "analytical_pipeline_diagram.png", dpi=300, bbox_inches="tight", facecolor="white")
plt.close()
print("Saved analytical pipeline diagram")
```
